# **Day 1: Introduction to Unsupervised Learning**

## **Meaning of Unsupervised Learning**
Unsupervised learning is a machine learning approach where the model is trained on data **without labeled outputs**.  
The goal is to **discover hidden patterns, structures, or relationships** directly from the data rather than predicting a known target.

The model answers questions like:
- What natural groups exist in the data?
- Which data points are similar or different?
- Is there an unusual or abnormal pattern?

---

## **Difference: Supervised vs Unsupervised Learning**

| **Aspect** | **Supervised Learning** | **Unsupervised Learning** |
|------|--------------------|----------------------|
| **Data** | Labeled data | Unlabeled data |
| **Objective** | Predict a known target | Discover hidden structure |
| **Output** | Class or value | Groups, patterns, embeddings |
| **Error signal** | Explicit (loss vs label) | Implicit (distance, density, variance) |
| **Examples** | Classification, Regression | Clustering, Dimensionality Reduction |
| **Evaluation** | Accuracy, RMSE, F1-score | Silhouette, inertia, domain insight |

In short:  
Supervised learning **predicts answers**, unsupervised learning **finds questions and structure**.

---

## **Real-World Applications of Unsupervised Learning**

- **Customer Segmentation**: Group customers based on behavior without predefined categories  
- **Anomaly Detection**: Detect fraud, system failures, or network intrusions  
- **Market Basket Analysis**: Identify frequently co-occurring products  
- **Dimensionality Reduction**: Compress high-dimensional data for visualization and modeling  
- **Document & Text Clustering**: Organize large text corpora without labels  
- **Feature Learning**: Extract meaningful representations before supervised training  
- **Recommendation Preprocessing**: Discover latent user or item groups  

---

## **Key Insight**
Unsupervised learning is often the **first step in real ML systems**, used to understand data before labels, models, or business rules exist.
