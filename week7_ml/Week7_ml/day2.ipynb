# **Day 2: Clustering Fundamentals**

## **What is Clustering**
Clustering is an unsupervised learning technique that groups data points such that **points within the same cluster are more similar to each other than to those in other clusters**.

The objective is not prediction, but **structure discovery** in unlabeled data.  
Clustering reveals natural groupings based purely on data geometry or density.

Common questions clustering answers:
- How many natural groups exist?
- Which data points behave similarly?
- Are there outliers or isolated patterns?

---

## **Similarity Measures**
Similarity measures define **how closeness between data points is computed**.  
Clustering quality depends directly on this choice.

### **Distance-Based Measures**
- **Euclidean Distance**: Straight-line distance; used by K-Means  
- **Manhattan Distance**: Sum of absolute differences; robust to outliers  
- **Minkowski Distance**: Generalized form of Euclidean and Manhattan  

### **Similarity-Based Measures**
- **Cosine Similarity**: Measures angle between vectors; common in text data  
- **Jaccard Similarity**: Measures overlap between sets; used for binary features  

### **Key Observations**
- Distance measures require **feature scaling**
- Different measures produce **different cluster shapes**
- No universal best similarity metric exists

---

## **Core Insight**
Clustering is meaningless without a well-chosen similarity measure; **distance defines structure**.
